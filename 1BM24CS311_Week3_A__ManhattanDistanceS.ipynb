```python

import heapq
def manhattan_distance(state, goal):
    distance = 0
    for i in range(9):
        tile = state[i]
        if tile != 0:
            goal_index = goal.index(tile)
            row1, col1 = i // 3, i % 3
            row2, col2 = goal_index // 3, goal_index % 3
            distance += abs(row1 - row2) + abs(col1 - col2)
    return distance
def astar(start, goal):
    pq = []
    h = manhattan_distance(start, goal)
    heapq.heappush(pq, (h, 0, start, [start]))
    visited = set()
    while pq:
        f, g, state, path = heapq.heappop(pq)
        if state == goal:
            return path
        if tuple(state) in visited:
            continue
        visited.add(tuple(state))
        zero = state.index(0)
        row, col = zero // 3, zero % 3
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]
        for dr, dc in moves:
            nr, nc = row + dr, col + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_zero = nr * 3 + nc
                new_state = state[:]
                new_state[zero], new_state[new_zero] = (
                    new_state[new_zero], new_state[zero]
                )
                if tuple(new_state) not in visited:
                    new_g = g + 1
                    new_h = manhattan_distance(new_state, goal)
                    new_f = new_g + new_h
                    heapq.heappush(
                        pq, (new_f, new_g, new_state, path + [new_state])
                    )
    return None
start = [1, 2, 3, 4, 0, 6, 7, 5, 8]
goal = [1, 2, 3, 4, 5, 6, 7, 8, 0]
path = astar(start, goal)
if path:
    for g, state in enumerate(path):
        h = manhattan_distance(state, goal)
        f = g + h
        print("State:")
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print("g(n) =", g)
        print("h(n) =", h)
        print("f(n) =", f)
        print("----------------")
    print("Total moves =", len(path) - 1)
else:
    print("No solution found")
```
